```python
# Function to clean the dataset
def clean_data(data):
    data = data.drop(columns=["Id"], errors="ignore")
    data = data.drop_duplicates()
    data = data.dropna()
    return data


# Function to calculate statistics
def calculate_statistics(data):
    print("Mean:\n", data.mean(numeric_only=True))
    print("\nMedian:\n", data.median(numeric_only=True))
    print("\nMinimum:\n", data.min(numeric_only=True))
    print("\nMaximum:\n", data.max(numeric_only=True))
    print("\nStandard Deviation:\n", data.std(numeric_only=True))


# Function to display dataset information
def display_data(data):
    print("Dataset Shape:", data.shape)
    print("\nDataset Preview:")
    print(data.head())


print("All functions defined successfully!")
```

```python
# Clean the original dataset using our function
cleaned_df = clean_data(pd.read_csv("Iris.csv"))

# Display dataset information
display_data(cleaned_df)

# Calculate statistics
print("\nSTATISTICAL ANALYSIS")
calculate_statistics(cleaned_df)
```

```python

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set a clean graph style
sns.set_theme(style="whitegrid")

# Create figure
plt.figure(figsize=(12, 7))

# Define species and colors
species_list = df['Species'].unique()
colors = ['#3498db', '#f39c12', '#2ecc71']

# Plot each species separately
for species, color in zip(species_list, colors):
    subset = df[df['Species'] == species]

    # Scatter plot
    plt.scatter(
        subset['SepalLengthCm'],
        subset['PetalLengthCm'],
        color=color,
        label=species.replace('Iris-', ''),
        s=75,
        alpha=0.8,
        edgecolor='white',
        linewidth=0.8
    )

    # Trend line
    z = np.polyfit(
        subset['SepalLengthCm'],
        subset['PetalLengthCm'],
        1
    )
    p = np.poly1d(z)

    x = np.linspace(
        subset['SepalLengthCm'].min(),
        subset['SepalLengthCm'].max(),
        100
    )

    plt.plot(x, p(x), color=color, linewidth=2.5)

# Titles and labels
plt.title(
    'Iris Flower Species Analysis',
    fontsize=19,
    fontweight='bold',
    pad=20
)

plt.xlabel('Sepal Length (cm)', fontsize=13, labelpad=10)
plt.ylabel('Petal Length (cm)', fontsize=13, labelpad=10)

# Legend outside graph
plt.legend(
    title='Flower Species',
    loc='upper left',
    bbox_to_anchor=(1.02, 1),
    frameon=True,
    fontsize=11
)

# Improve layout
plt.grid(True, linestyle='--', alpha=0.35)
plt.tight_layout()

# Display graph
plt.show()
```

The graph shows the relationship between sepal length and petal length for three Iris flower species. Iris-setosa has smaller petal lengths, while Iris-virginica generally has larger petal lengths. Iris-versicolor falls between the two. The trend lines show a positive relationship between sepal length and petal length across all three species.
